```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder,StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix,mean_absolute_error,mean_squared_error,r2_score
```

```python
data = pd.read_csv("autism_dataset.csv",na_values="?")
```

```python
print(data.isnull().sum())
```

```python
df = data.dropna()
```

```python
print(df.isnull().sum())
```

```python
df["Class/ASD"] = LabelEncoder().fit_transform(df["Class/ASD"])
```

```python
features = ['A1_Score','A2_Score','A3_Score','A4_Score','A5_Score',
            'A6_Score','A7_Score','A8_Score','A9_Score','A10_Score','age']
X = df[features]
y_class = df["Class/ASD"]
y_reg = df["result"]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(X_scaled, y_class, test_size=0.2, random_state=42)
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_scaled, y_reg, test_size=0.2, random_state=42)
```

```python
# DECISION TREE CLASSISIER :

from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
model.fit(X_train_cls, y_train_cls)
y_pred = model.predict(X_test_cls)
print("Decision Tree Accuracy :", accuracy_score(y_test_cls, y_pred))
print("Classification Report: ",classification_report(y_test_cls, y_pred))
print("Confusion Matrix: ",confusion_matrix(y_test_cls, y_pred))
```

```python
# DECISION TREE REGRESSOR  :

from sklearn.tree import DecisionTreeRegressor
model = DecisionTreeRegressor()
model.fit(X_train_reg, y_train_reg)
y_pred = model.predict(X_test_reg)
print("MAE :", mean_absolute_error(y_test_reg,y_pred))
print("MSE :", mean_squared_error(y_test_reg,y_pred))
print("R2 Score : ",r2_score(y_test_reg,y_pred))
```

```python
# RANDOM FOREST CLASSIFIER :

from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier()
model.fit(X_train_cls, y_train_cls)
y_pred = model.predict(X_test_cls)
print("Random Forest Accuracy:", accuracy_score(y_test_cls, y_pred))
print("Classification Report:",classification_report(y_test_cls, y_pred))
print("Confusion Matrix:",confusion_matrix(y_test_cls,y_pred))
```

```python
# RANDOM FOREST REGRESSOR :

from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor()
model.fit(X_train_reg, y_train_reg)
y_pred = model.predict(X_test_reg)
print("MAE :", mean_absolute_error(y_test_reg,y_pred))
print("MSE :", mean_squared_error(y_test_reg,y_pred))
print("R2 Score : ",r2_score(y_test_reg,y_pred))
```

```python
# K-NEAREST NEIGHBOR CLASSIFIER :

from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier()
model.fit(X_train_cls, y_train_cls)
y_pred = model.predict(X_test_cls)
print("Accuracy:", accuracy_score(y_test_cls, y_pred))
print(classification_report(y_test_cls, y_pred))
print("Confusion Matrix:",confusion_matrix(y_test_cls,y_pred))
```
